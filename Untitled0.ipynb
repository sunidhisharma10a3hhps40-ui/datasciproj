```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import statsmodels.api as sm
from statsmodels.stats.outliers_influence import variance_inflation_factor
import zipfile

# 1. Configure file path and targeted CSV inside zip
ZIP_FILE_PATH = "API_BX.KLT.DINV.WD.GD.ZS_DS2_en_csv_v2_1793.zip"
TARGET_CSV = "API_BX.KLT.DINV.WD.GD.ZS_DS2_en_csv_v2_1793.csv"

COUNTRY_COL = "Country Name"
YEAR_COL = "Year"
GDP_COL = "GDP_Growth"
INFLATION_COL = "Inflation"
FDI_COL = "FDI_Inflows"
TRADE_COL = "Trade_Pct_GDP"

# 2. Extract and load the correct CSV file from the ZIP archive (skipping World Bank header metadata rows)
with zipfile.ZipFile(ZIP_FILE_PATH) as z:
    with z.open(TARGET_CSV) as f:
        df = pd.read_csv(f, skiprows=4)

print(f"Dataset shape: {df.shape[0]} rows, {df.shape[1]} columns")
display(df.head())
df.info()
```

```python
# Identify year columns (which are purely numeric 4-digit strings)
year_cols = [col for col in df.columns if col.isdigit()]

# Melt the DataFrame from wide to long format
df_long = pd.melt(
    df,
    id_vars=['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code'],
    value_vars=year_cols,
    var_name='Year',
    value_name=FDI_COL
)

# Clean the 'Year' column to be integer type
df_long['Year'] = df_long['Year'].astype(int)

# Display the reshaped DataFrame
print(f"Reshaped dataset shape: {df_long.shape[0]} rows, {df_long.shape[1]} columns")
display(df_long.head())
df = df_long
```

```python
import os
import pandas as pd
import zipfile

# 1. Define the file names for your other indicator datasets
# (Please upload these files to your Google Colab file panel and update the paths if needed)
INDICATORS = {
    GDP_COL: {
        'zip': 'API_NY.GDP.MKTP.KD.ZG_DS2_en_csv_v2_1793.zip', # Default name for GDP Growth
        'csv': 'API_NY.GDP.MKTP.KD.ZG_DS2_en_csv_v2_1793.csv'
    },
    INFLATION_COL: {
        'zip': 'API_FP.CPI.TOTL.ZG_DS2_en_csv_v2_1793.zip',      # Default name for Inflation
        'csv': 'API_FP.CPI.TOTL.ZG_DS2_en_csv_v2_1793.csv'
    },
    TRADE_COL: {
        'zip': 'API_NE.TRD.GNFS.ZS_DS2_en_csv_v2_1793.zip',      # Default name for Trade % of GDP
        'csv': 'API_NE.TRD.GNFS.ZS_DS2_en_csv_v2_1793.csv'
    }
}

# Base keys for merging
join_keys = ['Country Name', 'Country Code', 'Year']

# df currently contains our reshaped FDI data.
# Let's ensure 'Year' is an integer in our base df first
df['Year'] = df['Year'].astype(int)
merged_df = df[['Country Name', 'Country Code', 'Year', FDI_COL]].copy()

# 2. Iterate through and merge each indicator
for col_name, files in INDICATORS.items():
    zip_path = files['zip']
    csv_path = files['csv']

    if os.path.exists(zip_path):
        print(f"Processing {col_name} from {zip_path}...")
        with zipfile.ZipFile(zip_path) as z:
            with z.open(csv_path) as f:
                # Load the dataset skipping World Bank metadata headers
                temp_df = pd.read_csv(f, skiprows=4)

        # Get year columns
        years = [col for col in temp_df.columns if col.isdigit()]

        # Melt wide dataset to long
        temp_long = pd.melt(
            temp_df,
            id_vars=['Country Name', 'Country Code'],
            value_vars=years,
            var_name='Year',
            value_name=col_name
        )
        temp_long['Year'] = temp_long['Year'].astype(int)

        # Merge with main dataframe
        merged_df = pd.merge(merged_df, temp_long, on=join_keys, how='outer')
    else:
        print(f"Warning: {zip_path} not found. Skipped merging {col_name}.")

# 3. Update active dataframe reference
df = merged_df
print(f"\nFinal Merged Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
display(df.head())
```

```python
import os
import pandas as pd
import zipfile

# Define the datasets we want to merge onto our existing df
INDICATORS_TO_MERGE = {
    'GDP_Growth': {
        'zip': 'API_NY.GDP.MKTP.KD.ZG_DS2_en_csv_v2_1793.zip',
        'csv': 'API_NY.GDP.MKTP.KD.ZG_DS2_en_csv_v2_1793.csv'
    },
    'Inflation': {
        'zip': 'API_FP.CPI.TOTL.ZG_DS2_en_csv_v2_1793.zip',
        'csv': 'API_FP.CPI.TOTL.ZG_DS2_en_csv_v2_1793.csv'
    },
    'Trade_Pct_GDP': {
        'zip': 'API_NE.TRD.GNFS.ZS_DS2_en_csv_v2_1793.zip',
        'csv': 'API_NE.TRD.GNFS.ZS_DS2_en_csv_v2_1793.csv'
    }
}

# Ensure our existing 'Year' column is integer-typed to match
df['Year'] = df['Year'].astype(int)

# We will keep adding columns to our existing dataframe 'df'
for col_name, paths in INDICATORS_TO_MERGE.items():
    zip_path = paths['zip']
    csv_path = paths['csv']

    if os.path.exists(zip_path):
        print(f"Extracting and merging {col_name} from {zip_path}...")
        with zipfile.ZipFile(zip_path) as z:
            with z.open(csv_path) as f:
                temp_df = pd.read_csv(f, skiprows=4)

        # Melt the new indicator dataset
        year_cols = [col for col in temp_df.columns if col.isdigit()]
        temp_long = pd.melt(
            temp_df,
            id_vars=['Country Name', 'Country Code'],
            value_vars=year_cols,
            var_name='Year',
            value_name=col_name
        )
        temp_long['Year'] = temp_long['Year'].astype(int)

        # Merge with main dataframe on Country Name, Country Code, and Year
        df = pd.merge(
            df,
            temp_long[['Country Name', 'Country Code', 'Year', col_name]],
            on=['Country Name', 'Country Code', 'Year'],
            how='outer'
        )
    else:
        print(f"File '{zip_path}' not found in directory. Skipping {col_name}.")

print(f"\nMerged dataframe shape: {df.shape[0]} rows, {df.shape[1]} columns")
display(df.head())
```

```python
# Define the target indicators we want to analyze
target_cols = [GDP_COL, INFLATION_COL, FDI_COL, TRADE_COL]

# Filter to only the columns that actually exist in our DataFrame
numeric_cols = [col for col in target_cols if col in df.columns]

print(f"Analyzing available columns: {numeric_cols}\n")

if len(numeric_cols) > 0:
    # 1. Check missing values before cleaning
    print("Missing values before imputation:\n", df[numeric_cols].isnull().sum())

    # 2. Country-wise median imputation (falls back to overall median if a country has all NaNs)
    for col in numeric_cols:
        df[col] = df.groupby(COUNTRY_COL)[col].transform(lambda x: x.fillna(x.median()) if not x.isnull().all() else x)
        df[col] = df[col].fillna(df[col].median())

    print("\nMissing values after imputation:\n", df[numeric_cols].isnull().sum())

    # 3. Summary statistics with IQR
    summary = df[numeric_cols].describe().T
    summary["IQR"] = summary["75%"] - summary["25%"]
    display(summary[["count", "mean", "50%", "std", "IQR", "min", "max"]])
else:
    print("Error: None of the target numeric columns are present in the DataFrame.")
```
